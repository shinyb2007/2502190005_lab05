```python
raw_lines = [
    "MS-FLG-2026-0041, 24.98",
    " ms-flg-2026-0042, 25.01 ",
    "AL-BRK-2026-0007, 24.95",
    "MS-FLG-2026-0043, 25.04",
    "AL-BRK-2026-0008, 25.12",
    "ms-shf-2026-0101, 24.88",
    "MS-FLG-2026-0044, 24.99",
    "AL-BRK-2026-0009, 25.02",
    "MS-SHF-2026-0102, 25.07",
    "MS-FLG-2026-0045, 24.93"
]

codes = []
dias = []

for line in raw_lines:
    code_part, dia_part = line.split(",")
    codes.append(code_part.strip().upper())
    dias.append(float(dia_part.strip()))

print(len(codes), "parts registered")
print(codes[0], dias[0])
print(len(codes) == len(dias))
```

```python
families = ["FLG", "BRK", "SHF"]

for family in families:
    values = []

    for i in range(len(dias)):
        if codes[i].split("-")[1] == family:
            values.append(dias[i])

    print("\n", family)
    print("Count:", len(values))
    print("Mean:", round(sum(values) / len(values), 3))
    print("Min:", min(values))
    print("Max:", max(values))
    print("Spread:", round(max(values) - min(values), 3))

print("\nOUT OF TOLERANCE:")
out = 0

for i in range(len(dias)):
    if not (24.95 <= dias[i] <= 25.05):
        print(codes[i], dias[i])
        out += 1

print("\nOverall:")
print("Total parts:", len(dias))
print("Total out of tolerance:", out)
print("Acceptance:", round((len(dias) - out) / len(dias) * 100, 1), "%")
```

```python
backup = dias[:]

cleaned = []

for value in dias:
    if 24.95 <= value <= 25.05:
        cleaned.append(value)

print("Original length:", len(dias))
print("Backup length:", len(backup))
print("Cleaned length:", len(cleaned))
print("Original unchanged:", dias)

backup = dias
backup.append(99.9)
print("New changed:",dias)
```
